**Importation des données**

```python
import pandas as pd
from IPython.display import display

df1 = pd.read_excel("Ventes_Février_2025.xlsx")
df2 = pd.read_excel("Ventes_Janvier_2025.xlsx")
df3 = pd.read_excel("Ventes_Mars_2025.xlsx")
df4 = pd.read_excel("Referentiels_Projet.xlsx")
```

**Analyse et Nettoyage**

```python
display(df1.isnull().sum())
```

```python
lignes_manquante = df1[df1.isnull().any(axis=1)]

display(lignes_manquante)
```

```python
mask_qte = (
    df1["Quantité"].isna()
   &df1["Montant"].notna()
   &df1["Prix_Unitaire"].notna()
   &df1["Remise"].notna()
)

df1.loc[mask_qte, "Quantité"] = (
    df1.loc[mask_qte, "Montant"] /(
    df1.loc[mask_qte, "Prix_Unitaire"]*
    (1-df1.loc[mask_qte, "Remise"])
    )
)
```

```python
df1["Région"] = df1["Région"].fillna("Auvergne-Rhône-Alpes")
```

```python
df1["Produit"] = df1["Produit"].fillna("Produit F")
```

```python
df1["Client_ID"] = df1["Client_ID"].fillna("C0063")
```

```python
display(df1.isnull().sum())
```

```python
df1.to_excel("Ventes_Février_2025_clean.xlsx", index=False)
```
